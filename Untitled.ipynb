```python
author_styles = [
    "William Shakespeare. His writing style is marked by rich language, intricate wordplay, and deep emotional exploration. He often used iambic pentameter, blending prose and poetry, with vivid imagery and rhetorical devices like soliloquies to reveal character thoughts.",
    "Jane Austen. Her style is characterized by precise prose, social commentary, and nuanced character development. She employs a formal, elegant style, often using dialogue and observation to reveal character and relationships. Her narrative focuses on manners and social conventions, utilizing irony and subtle humor to create compelling stories.",
    "Charles Dickens. His work is renowned for its vivid imagery, social realism, and focus on everyday life. He employs a distinctive, melodramatic style, often using exaggerated language and dramatic situations to engage the reader. His stories explore themes of poverty, injustice, and redemption, often relying on memorable characters and memorable plotlines.",
    "Leo Tolstoy. His writing style is known for its expansive scope, psychological depth, and moral complexity. He employs a long, flowing prose style that prioritizes character development and thematic exploration. His narrative often delves into the inner lives of his characters, utilizing detailed descriptions and philosophical reflections.",
    "F. Scott Fitzgerald. His writing style is marked by a nostalgic tone, themes of disillusionment, and a focus on the American Dream. He utilizes a sophisticated, often melancholic prose style, often employing imagery to evoke a sense of longing and decay. His narratives often portray the complexities of wealth and social status.",
    "Ernest Hemingway. His style is characterized by short, declarative sentences, minimalist prose, and an emphasis on action and dialogue. He employs a precise and understated style, prioritizing imagery and implication over explicit description. His stories often explore themes of masculinity, war, and loss.",
    "Virginia Woolf. Her style is known for its stream-of-consciousness technique, psychological exploration, and use of imagery. She employs a relatively short, flowing prose style, prioritizing the internal experience of her characters. Her narratives often explore themes of memory, perception, and identity.",
    "Jane Greene. Her writing is distinct for its witty dialogue, complex psychological portraits, and exploration of social issues. She utilizes a dynamic, conversational style with subtle social satire. Her narratives are often subtle and nuanced, prioritizing character relationships and the unfolding of complex situations."
]
```

Needs an embedding model

```python
from ollama import embed
```

```python
# Generate an embedding for a batch input
response = embed(model='nomic-embed-text', input=author_styles)
# Access the embedding vectors
embedding = response['embeddings']

print(len(embedding), len(embedding[0]), len(embedding[0]))
```

```python
#pip install numpy
import numpy as np

vectors = np.array(embedding)

query = np.array(embed(model='nomic-embed-text', input='I want suspense and thrill.')['embeddings'])

# Compute Euclidean distances
differences = vectors - query  # shape: (n_vectors, vector_dim)
distances = np.linalg.norm(differences, axis=1)

# Index of closest vector
closest_index = np.argmin(distances)

# Sorted indices by distance
sorted_indices = np.argsort(distances)

print("Closest vector index:", closest_index)
print("Distance:", distances[closest_index])
print("Sorted indices by distance:", sorted_indices)
```

```python
author_styles[closest_index]
```

```python
# Sort author_styles according to the sorted_indices
sorted_author_styles = [author_styles[i] for i in sorted_indices]

# Optionally print it
for num, talk in enumerate(sorted_author_styles):
    print(num+1, talk)
    print()
```

```python
retrieved_author_styles = sorted_author_styles[:3]
```

```python
from IPython.display import Markdown, display

retrieved_content = '\n\n'.join(retrieved_author_styles)

chat_response = chat(model='gemma3:1b', messages=[
    {
    'role': 'user',
    'content': 'User Question:\nWhich author writes the most thrilling stories?\n\n' + 'Answer the user question on author styles as best as you can based on the retrieved author styles.\n\nRetrieved content:\n\n' + retrieved_content,
    },
])

display(Markdown(chat_response.message.content))
```
